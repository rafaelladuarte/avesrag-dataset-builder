```python
import pandas as pd
```

```python
path_zenodo = 'data/raw/2021.07.26 CBRO 2021 Zenodo Release.xlsx'
```

```python
df_zenodo = pd.read_excel(
    path_zenodo,sheet_name="Lista Primária"
)
```

```python
df_especie = df_zenodo[df_zenodo["Categoria"]=="Espécie"]
df_especie.head()
```

```python
df_especie.info()
```

```python
df_especie_new = df_especie[
    [
        "Nome do táxon (sem autoria)",
        "Ordem",
        "Família",
        "Gênero",
        "Espécie",
        "Nome em Português",
        "English Name",
        "Status"
    ]
]
df_especie_new
```

```python
df_especie_new = df_especie_new.rename(
    columns={
        "Nome do táxon (sem autoria)": "taxonomia",
        "Ordem": "ordem",
        "Família": "familia",
        "Gênero": "genero",
        "Espécie": "especie",
        "Nome em Português": "nome_pt",
        "English Name": "nome_en",
        "Status": "staus_cbro"
    },
)
df_especie_new
```

```python
df_especie_new.to_json(
    "data/interim/zenodo_bird_brasil.json",
    orient="records",
    force_ascii=False
)
```
