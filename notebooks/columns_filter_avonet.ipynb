```python
import pandas as pd
```

```python
path_avonet =  "data/raw/AVONET Supplementary dataset 1.xlsx"
```

```python
df_avonet = pd.read_excel(
    path_avonet,sheet_name="AVONET2_eBird"
)
df_avonet.head()
```

```python
df_avonet = df_avonet[
    [
        "Species2",
        "Beak.Length_Culmen",
        "Beak.Width",	
        "Beak.Depth",	
        "Tarsus.Length",	
        "Wing.Length",	
        "Tail.Length",
        "Mass",
        "Habitat",	
        "Habitat.Density",	
        "Migration",
        "Trophic.Level",	
        "Trophic.Niche",
        "Primary.Lifestyle"
    ]
]
df_avonet
```

```python
df_avonet_new = df_avonet.rename(
    columns={
        "Species2": "taxonomia",
        "Beak.Length_Culmen": "comprimento_bico_mm",
        "Beak.Width": "largura_bico_mm",	
        "Beak.Depth": "profundidade_bico_mm",	
        "Tarsus.Length": "comprimento_tarso_mm",	
        "Wing.Length": "comprimento_asa_mm",	
        "Tail.Length": "comprimento_causa_mm",
        "Mass": "massa_corporal_g",
        "Habitat": "bioma_habitat",	
        "Habitat.Density": "densidade_habitat",	
        "Migration": "migracao",
        "Trophic.Level": "nivel_trofico",	
        "Trophic.Niche": "nicho_trofico",
        "Primary.Lifestyle": "nicho_locomotor"
    },
)
df_avonet_new
```

```python
density_map = {
    1: "denso",
    2: "semi-aberto",
    3: "aberto"
}

df_avonet_new["densidade_habitat"] = df_avonet_new ["densidade_habitat"].map(density_map)
df_avonet_new
```

```python
migration_map = {
    1: "sedentaria",
    2: "parcialmente migratoria",
    3: "migratoria"
}

df_avonet_new["migracao"] = df_avonet_new ["migracao"].map(migration_map)
df_avonet_new
```

```python
habitat_map = {
    "Desert": "deserto",
    "Rock": "rochoso",
    "Grassland": "campo",
    "Shrubland": "arbustivo",
    "Woodland": "bosque",
    "Forest": "floresta",
    "Human modified": "ambiente antropico",
    "Wetland": "area umida",
    "Riverine":  "ribeirinho",
    "Coastal": "costeiro",
    "Marine": "marinho"
}

df_avonet_new["bioma_habitat"] = df_avonet_new ["bioma_habitat"].map(habitat_map)
df_avonet_new
```

```python
nivel_trofico_map = {
    "Herbivore": "herbivora",
    "Carnivore": "carnivora",
    "Scavenger": "necrofaga",
    "Omnivore": "onivora"
}

df_avonet_new["nivel_trofico"] = df_avonet_new ["nivel_trofico"].map(nivel_trofico_map)
df_avonet_new
```

```python
nicho_trofico_map = {
    "Frugivore": "frugivora",
    "Granivore": "granivora",
    "Nectarivore": "nectarivora",
    "Herbivore": "herbivora",
    "Herbivore aquatic": "herbivora aquatica",
    "Invertivore": "invertivora",
    "Vertivore": "vertivora",
    "Aquatic Predator": "predadora aquatica",
    "Scavenger": "necrofaga",
    "Omnivore": "onivora"
}

df_avonet_new["nicho_trofico"] = df_avonet_new ["nicho_trofico"].map(nicho_trofico_map)
df_avonet_new
```

```python
nicho_locomotor_map =  {
    "Aerial": "aerea",
    "Terrestrial": "terrestre",
    "Insessorial": "insessorial",
    "Aquatic": "aquatica",
    "Generalist": "generalista"
}

df_avonet_new["nicho_locomotor"] = df_avonet_new ["nicho_locomotor"].map(nicho_locomotor_map)
df_avonet_new
```

```python
df_avonet_new.to_json(
    "data/interim/avonet_bird_morphology_global.json",
    orient="records",
    force_ascii=False
)
```
