```python
import pandas as pd
import re
```

```python
def tem_caracter_chines(texto):
    # Regex para caracteres chineses (CJK Unified Ideographs)
    padrao = r'[\u4e00-\u9fff]'
    return bool(re.search(padrao, texto))
```

```python
path = 'data/treat/dedup2_result_scraper_wikiaves_udi.json'
df = pd.read_json(path)
df
```

```python
df['contem_chines'] = df['resumo_llm'].apply(tem_caracter_chines)
df[df['contem_chines'] == True]["resumo_llm"]
```

```python
df["tamanho"].unique()
```

```python
df.loc[df["tamanho"]=='Grande','tamanho'] = 'grande'
df.loc[df["tamanho"]=='Médio','tamanho'] = 'media'
df.loc[df["tamanho"]=='médio','tamanho'] = 'media'
df.loc[df["tamanho"]=='Pequeno','tamanho'] = 'pequena'
df.loc[df["tamanho"]=='pequeno','tamanho'] = 'pequena'
```

```python
df["tipo_bico"].unique()
```

```python
new_df = df[[
    'taxonomia', 'ordem', 'familia', 'genero', 'especie',
    'nome_pt', 'nome_eng', 'url_wikiaves', 'url_image', 'resumo_llm',
    'tamanho', 'tipo_bico', 'cores', 'habitat','dieta_principal',
    'atividade'
]]
new_df
```
