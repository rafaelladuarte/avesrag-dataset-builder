````python
import pandas as pd
import numpy as np

import json
import ast
import re
````

````python
path_json = "result_llm_wikiaves_udi.json"
df = pd.read_json(path_json)
df.head()
````

````python
def string_to_dict(string_input):
    try:
        match = re.search(r'```json\s*(.*?)\s*```', string_input, re.DOTALL)

        content = match.group(1)

        string_clean = re.sub(r'^```json\n|\n```$','',content.strip())
        string_clean = string_clean.replace('\\','')
        string_clean = string_clean.replace('...','')

        try:
            string_json = json.loads(string_clean)
        except json.JSONDecodeError:
            string_json = ast.literal_eval(string_clean)
        
        return string_json
    except:
        return {}
````

````python
examples_string = [
    "```json\n{'locomocao': 'voadora', 'sociabilidade': 'solitária', 'atividade': ['diurna']}\n```",
    "```json\n{'bico': 'Pescador', 'asa': 'Pernalta', 'pata': 'Palmada', 'cores': [\"negro\", \"branco\", \"creme-acinzentada\"], 'tamanho': 'Grande'} \n```",
    "```json\n{'habitat': []}\n```",
    "```json\n{'locomocao': '...', 'sociabilidade': '...', 'atividade': ['noturna']}\n```",
    "```json\n{'bico': 'Insetívoro', 'asa': 'Elíptica', 'pata': '', 'cores': [], 'tamanho': 'Pequeno'}\n```ariatric  "
]
````

````python
bico_by_familly = {
    "Generalista": ["Turdidae", "Thraupidae", "Mimidae"],
    "Insetívoro": ["Tyrannidae", "Troglodytidae", "Furnariidae"],
    "Granívoro": ["Fringillidae", "Emberizidae", "Cardinalidae"],
    "Comedor de sementes de pinho": [],
    "Nectívoro": ["Trochilidae", "Meliphagidae"],
    "Frugívoro": ["Ramphastidae", "Psittacidae", "Cotingidae"],
    "Insetos em troncos": ["Picidae"],
    "Rede de pesca": ["Sulidae"],
    "Pescador de superfície": ["Laridae", "Sternidae"],
    "Limícola": ["Scolopacidae", "Charadriidae"],
    "Sondador": ["Recurvirostridae", "Ibidorhynchidae"],
    "Filtrador": ["Phoenicopteridae", "Anatidae"],
    "Pescador": ["Alcedinidae", "Anhingidae"],
    "Pato mergulhador": ["Anatidae"],
    "Carniceiro": ["Cathartidae"],
    "Raptorial": ["Accipitridae", "Falconidae"]
}


def recap_bico_by_familly(familly: str):
    for key, values in bico_by_familly.items():
        if familly in values:
            return key
    return None
````

````python
df_graph["tipo_bico_recap"] = df_graph["familia"].apply(recap_bico_by_familly)
df_graph["tipo_bico_recap"]
````
