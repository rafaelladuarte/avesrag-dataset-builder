## Merge dedup + resume + parameterizes

```python
df_par.to_json("../data/treat/parameterized_dataset_udi.json", orient="records")
```

```python
df_dedup_result = pd.read_json("../data/treat/dedup_result_scraper_wikiaves_udi.json")
df_par = pd.read_json("../data/treat/new_parameterized_dataset_udi.json")
```

```python
df_dedup_result.info()
```

```python
df_dedup_result = df_dedup_result.drop(columns=["Unnamed: 0",""])
df_dedup_result
```

```python
df_par.info()
```

```python
df_left_join = pd.merge(df_dedup_result, df_par, on="taxonomia", how="left")
df_left_join
```

```python
df_left_join.info()
```

```python
df_left_join.columns
```

```python
df_left_join = df_left_join[['taxonomia', 'ordem_x', 'familia_x', 'subfamilia_x', 'genero_x',
       'especie_x', 'nome_pt_x', 'nome_eng_x', 'url_wikiaves_x',
       'caracteristicas_x', 'alimentacao_x', 'reproducao_x', 'habitos_x',
       'dist_geo_x', 'resumo_llm_x', "url_image_x",'formato_asa', 'tipo_pata', 'cores',
       'tamanho', 'caracteristica_trofica', 'dieta_principal', 'habitat',
       'locomocao', 'sociabilidade', 'atividade', 'tipo_bico']]
```

```python
df_left_join = df_left_join.rename(
    columns={
        'ordem_x': 'ordem',
        'familia_x': 'familia',
        'subfamilia_x': 'subfamilia',
        'genero_x': 'genero',
        'especie_x': 'especie',
        'nome_pt_x': 'nome_pt',
        'nome_eng_x': 'nome_eng',
        'url_wikiaves_x': 'url_wikiaves',
        'caracteristicas_x': 'caracteristicas',
        'alimentacao_x': 'alimentacao',
        'reproducao_x': 'reproducao',
        'habitos_x': 'habitos',
        'dist_geo_x': 'dist_geo',
        'resumo_llm_x' : 'resumo_llm',
        "url_image_x": 'url_image',
    }
)
df_left_join
```

```python
df_left_join = df_left_join.replace(["", None], np.nan)
df_left_join
```

```python
df_left_join.to_json("../data/treat/new_parameterized_dataset_udi.json", orient="records")
```
